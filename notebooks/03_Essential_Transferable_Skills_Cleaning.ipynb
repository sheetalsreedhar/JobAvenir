Step 1: Load both the datasets

```python
import pandas as pd
from pathlib import Path

# Main project folder
project_folder = Path(
    r"C:\Users\belie\OneDrive\Desktop\ADT MySQL\Lab 7 Project Milestone\JobAvenir"
)

# O*NET data folder
onet_folder = project_folder / "data" / "Onet"

# Load Essential Skills
essential_skills = pd.read_csv(
    onet_folder / "essential_skills.csv",
    dtype="string",
    encoding="utf-8-sig"
)

# Load Transferable Skills
transferable_skills = pd.read_csv(
    onet_folder / "transferable_skills.csv",
    dtype="string",
    encoding="utf-8-sig"
)

# Display dataset information
print("Essential Skills:", essential_skills.shape)
print("Transferable Skills:", transferable_skills.shape)

print("\nEssential Skills Columns:")
print(essential_skills.columns.tolist())

print("\nTransferable Skills Columns:")
print(transferable_skills.columns.tolist())

display(essential_skills.head())
display(transferable_skills.head())
```

Step 2: Checking missing values and duplicates

```python
datasets = {
    "Essential Skills": essential_skills,
    "Transferable Skills": transferable_skills
}

for name, df in datasets.items():

    print("\n==============================")
    print("Dataset:", name)
    print("==============================")

    print("\nRows and Columns:")
    print(df.shape)

    print("\nColumn Names:")
    print(df.columns.tolist())

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\nFirst Five Records:")
    display(df.head())
```

Step 3: Clean Essential Skills

```python
# Make a copy of the original dataset
essential_clean = essential_skills.copy()

# Rename columns
essential_clean = essential_clean.rename(columns={
    "O*NET-SOC Code": "occupation_id",
    "Title": "occupation_title",
    "Element ID": "element_id",
    "Element Name": "skill_name",
    "Scale ID": "scale_id",
    "Scale Name": "scale_name",
    "Data Value": "data_value",
    "N": "sample_size",
    "Standard Error": "standard_error",
    "Lower CI Bound": "lower_ci_bound",
    "Upper CI Bound": "upper_ci_bound",
    "Recommend Suppress": "recommend_suppress",
    "Not Relevant": "not_relevant",
    "Date": "date_updated",
    "Domain Source": "domain_source"
})

# Remove extra spaces from text columns
for column in essential_clean.columns:

    essential_clean[column] = (
        essential_clean[column]
        .astype("string")
        .str.strip()
    )

# Replace empty strings with missing values
essential_clean = essential_clean.replace(
    r"^\s*$",
    pd.NA,
    regex=True
)

# Convert numerical columns
numeric_columns = [
    "data_value",
    "sample_size",
    "standard_error",
    "lower_ci_bound",
    "upper_ci_bound"
]

for column in numeric_columns:

    essential_clean[column] = pd.to_numeric(
        essential_clean[column],
        errors="coerce"
    )

# Standardize Y/N values
for column in ["recommend_suppress", "not_relevant"]:

    essential_clean[column] = (
        essential_clean[column]
        .str.upper()
    )

# Remove completely identical duplicate records
essential_clean = essential_clean.drop_duplicates()

# Display cleaned results
print("Cleaned Essential Skills Records:", len(essential_clean))

print("\nMissing Values:")
print(essential_clean.isnull().sum())

print("\nDuplicate Rows:")
print(essential_clean.duplicated().sum())

display(essential_clean.head(10))
```

```python
# Make a copy
transferable_clean = transferable_skills.copy()

# Rename columns
transferable_clean = transferable_clean.rename(columns={
    "O*NET-SOC Code": "occupation_id",
    "Title": "occupation_title",
    "Element ID": "element_id",
    "Element Name": "skill_name",
    "Scale ID": "scale_id",
    "Scale Name": "scale_name",
    "Data Value": "data_value",
    "N": "sample_size",
    "Standard Error": "standard_error",
    "Lower CI Bound": "lower_ci_bound",
    "Upper CI Bound": "upper_ci_bound",
    "Recommend Suppress": "recommend_suppress",
    "Not Relevant": "not_relevant",
    "Date": "date_updated",
    "Domain Source": "domain_source"
})

# Remove extra spaces
for column in transferable_clean.columns:

    transferable_clean[column] = (
        transferable_clean[column]
        .astype("string")
        .str.strip()
    )

# Replace empty strings with missing values
transferable_clean = transferable_clean.replace(
    r"^\s*$",
    pd.NA,
    regex=True
)

# Convert numerical columns
numeric_columns = [
    "data_value",
    "sample_size",
    "standard_error",
    "lower_ci_bound",
    "upper_ci_bound"
]

for column in numeric_columns:

    transferable_clean[column] = pd.to_numeric(
        transferable_clean[column],
        errors="coerce"
    )

# Standardize Y/N values
for column in ["recommend_suppress", "not_relevant"]:

    transferable_clean[column] = (
        transferable_clean[column]
        .str.upper()
    )

# Remove completely identical duplicate records
transferable_clean = transferable_clean.drop_duplicates()

# Display cleaned results
print("Cleaned Transferable Skills Records:", len(transferable_clean))

print("\nMissing Values:")
print(transferable_clean.isnull().sum())

print("\nDuplicate Rows:")
print(transferable_clean.duplicated().sum())

display(transferable_clean.head(10))
```

Step 5: Validate both datasets

```python
# Load your existing cleaned Occupation Data
occupation_clean = pd.read_csv(
    project_folder / "data" / "cleaned" / "occupations_clean.csv",
    dtype="string"
)

# Get valid occupation IDs
valid_occupations = set(
    occupation_clean["occupation_id"].dropna().str.strip()
)

# Validate each dataset
for name, df in {
    "Essential Skills": essential_clean,
    "Transferable Skills": transferable_clean
}.items():

    print("\nDataset:", name)

    # Check occupation IDs
    invalid_occupations = df[
        ~df["occupation_id"].isin(valid_occupations)
    ]

    print(
        "Invalid Occupation References:",
        len(invalid_occupations)
    )

    # Check composite keys
    duplicate_keys = df.duplicated(
        subset=[
            "occupation_id",
            "element_id",
            "scale_id"
        ]
    ).sum()

    print("Duplicate Rating Keys:", duplicate_keys)

    # Check missing essential identifiers
    required_columns = [
        "occupation_id",
        "element_id",
        "skill_name",
        "scale_id"
    ]

    print("\nMissing Required Values:")
    print(df[required_columns].isnull().sum())

    # Check invalid numeric values that may require review
    print("\nMissing Rating Values:")
    print(df["data_value"].isnull().sum())
```

Step 6: Save both cleaned CSV files

```python
# Output folder
output_folder = project_folder / "data" / "cleaned"

output_folder.mkdir(
    parents=True,
    exist_ok=True
)

# Save cleaned Essential Skills
essential_clean.to_csv(
    output_folder / "essential_skills_clean.csv",
    index=False,
    encoding="utf-8-sig"
)

# Save cleaned Transferable Skills
transferable_clean.to_csv(
    output_folder / "transferable_skills_clean.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Both cleaned datasets saved successfully!")

print("\nEssential Skills:", len(essential_clean))
print("Transferable Skills:", len(transferable_clean))

print("\nFiles saved in:")
print(output_folder)
```
